```sql
CREATE TABLE IF NOT EXISTS data_governance.gold_access.fact_user_access_activity (
    -- 🔑 EVENT CORE
    event_id STRING,
    event_timestamp TIMESTAMP,
    event_time STRING,
    event_date DATE,
    version STRING,
    audit_level STRING,

    -- 🏢 ORG
    account_id STRING,
    workspace_id STRING,
    workspace_id_req STRING,
    metastore_id STRING,

    -- 👤 USER
    user_email STRING,
    user_subject_name STRING,
    run_by STRING,
    run_as STRING,
    run_by_display_name STRING,
    run_as_display_name STRING,
    is_system_user BOOLEAN,

    -- 🔐 ACCESS
    service_name STRING,
    action_name STRING,
    catalog_name STRING,
    schema_name STRING,
    acting_resource STRING,
    include_browse BOOLEAN,

    -- 🌐 TECH CONTEXT
    client_type STRING,
    source_ip_address STRING,
    is_internal_ip BOOLEAN,
    user_agent STRING,
    session_id STRING,

    -- 📊 RESULT
    status_code INT,
    error_message STRING,
    response_result STRING,
    is_success BOOLEAN,
    max_results INT,
    request_id STRING,

    -- ⏱ PARTITION
    event_year INT,
    event_month INT,
    event_day INT,
    event_hour INT,

    -- 🧾 PIPELINE
    load_timestamp TIMESTAMP
)
USING DELTA
PARTITIONED BY (event_year, event_month, event_day);
```
